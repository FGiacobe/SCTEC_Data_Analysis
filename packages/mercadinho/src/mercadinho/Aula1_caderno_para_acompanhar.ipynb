# 🧮 Bloco 1 — Numpy: contas em massa

## O que é o Numpy?
NumPy (Numerical Python) é a biblioteca do Python para fazer contas com MUITOS números de uma vez, e muito rápido.

- 🔢 **Contas em massa**: Somar, multiplicar ou tirar a média de milhares de números com uma única linha.
- ⚡ **Muito rápido**: Por baixo, usa código em C, bem mais veloz que um laço for em Python.
- 🧱 **Base de tudo**: O Pandas, o Matplotlib e o scikit-learn são construídos em cima do Numpy.

💡  Para o analista: o Numpy é o "motor" por baixo do Pandas. Entender o Numpy ajuda a entender o Pandas.

## Por que o Numpy é mais rápido que uma lista?
Uma analogia para guardar: carros espalhados × um trem.

**Lista do Python:** cada número fica num lugar diferente da memória, como carros espalhados num estacionamento

**Array do Numpy:** todos os números ficam lado a lado e são do mesmo tipo, como os vagões de um trem

💡  O computador anda muito mais rápido num trem do que buscando carro por carro. E o Numpy ainda usa código em C por baixo.

## Na prática: quanto mais rápido?
Vamos medir: dobrar 1 milhão de números com uma lista e com um array.

💡  Os tempos variam de computador para computador, mas o array é sempre muito mais rápido. Com poucos números tanto faz; com milhões, faz toda a diferença.

```python
from time import perf_counter

n = 1_000_000
lista = list(range(n))
array = np.arange(n)

t0 = perf_counter()
dobro_lista = [x * 2 for x in lista]
t1 = perf_counter()
dobro_array = array * 2
t2 = perf_counter()

print('lista:', round(t1 - t0, 4), 's')
print('numpy:', round(t2 - t1, 4), 's')
print('numpy é', round((t1 - t0) / (t2 - t1)), 'vezes mais rápido')
```

## O que é um array?
Um array é uma "lista turbinada": guarda vários valores numa única variável, todos do MESMO tipo, e faz contas com todos de uma vez.

💡  lista * 2 REPETE a lista · array * 2 MULTIPLICA cada elemento. É a diferença mais importante para quem está começando.

```python
lista = [1, 2, 3]
array = np.array([1, 2, 3])

print(lista * 2)
print(array * 2)
print(type(lista))
print(type(array))
print(array.dtype)
```

## Conhecendo um array: shape, ndim, size, dtype
Todo array tem "características" que você consulta, sem parênteses:

💡  shape = (linhas, colunas) · ndim = nº de dimensões · size = total de elementos · dtype = tipo dos elementos · len = nº de linhas

```python
vendas = np.array([[120, 150, 130, 160],
                   [90, 110, 100, 130],
                   [70, 80, 95, 85]])

print(vendas.shape)
print(vendas.ndim)
print(vendas.size)
print(vendas.dtype)
print(len(vendas))
```

## Dimensões: 1D, 2D e 3D
O shape diz quantas dimensões existem e o tamanho de cada uma:

- 1D **Vetor · (5,)**: Uma linha de valores. Ex.: temperatura de 5 dias, ou os preços dos produtos.
- 2D **Matriz · (3, 4)**: Linhas × colunas, como uma planilha. Ex.: 3 lojas × 4 semanas de vendas.
- 3D **Tensor · (2, 3, 4)**: Camadas de tabelas. Ex.: alunos × matérias × bimestres, ou uma imagem colorida.

💡  Pense nas dimensões como: 1D = uma linha · 2D = uma tabela · 3D = uma pilha de tabelas.

## Um array tem um tipo só
Todos os elementos de um array são do MESMO tipo. Se misturar, o Numpy escolhe um tipo que sirva para todos:

💡  int64 = inteiros · float64 = decimais · <U = texto · astype() converte o tipo (e devolve um array novo).

```python
misto = np.array([1, 2.5, 3])
print(misto, misto.dtype)

inteiros = np.array([1, 2, 3])
print(inteiros.dtype)

como_float = inteiros.astype(float)
print(como_float, como_float.dtype)

texto = np.array([1, 'a', 3])
print(texto, texto.dtype)
```

## Posições de um array: contando a partir do zero
Cada valor tem um "endereço": a posição (índice).

|  | 1º | 2º | 3º | 4º | 5º |
|---|---|---|---|---|---|
| Posição | 0 | 1 | 2 | 3 | 4 |
| Posição negativa | -5 | -4 | -3 | -2 | -1 |
| Valor (preço) | 28.9 | 8.5 | 5.2 | 18.9 | 5.8 |

⚠️  A contagem começa em ZERO. O -1 é sempre o último elemento, sem precisar saber o tamanho.

## Indexação: pegando um elemento
Para pegar UM elemento, use colchetes com a posição. Também dá para trocar um valor:

💡  array[posição] pega um elemento · array[posição] = novo troca o valor · posição negativa conta do fim.

```python
precos = np.array([28.9, 8.5, 5.2, 18.9, 5.8])

print(precos[0])
print(precos[2])
print(precos[-1])
print(precos[-2])

precos[1] = 9.0
print(precos)
```

## Slicing: pegando uma fatia
Slicing pega vários elementos de uma vez, com [início:fim:passo]. O fim NÃO entra:

💡  [1:4] → posições 1, 2 e 3 · [:3] → do começo até a 2 · [2:] → da 2 até o fim · [::2] → de 2 em 2 · [::-1] → invertido

```python
precos = np.array([28.9, 8.5, 5.2, 18.9, 5.8])

print(precos[1:4])
print(precos[:3])
print(precos[2:])
print(precos[::2])
print(precos[::-1])
```

## Indexação e slicing em 2D
Em arrays 2D usamos [linha, coluna]. O dois-pontos (:) significa "todos":

💡  Primeiro a LINHA, depois a COLUNA · mat[0, :] = toda a linha 0 · mat[:, 1] = toda a coluna 1

```python
mat = np.array([[120, 150, 130, 160],
                [90, 110, 100, 130],
                [70, 80, 95, 85]])

print(mat[0, 0])
print(mat[1, 2])
print(mat[2, -1])
print(mat[0, :])
print(mat[:, 1])
print(mat[0:2, :])
print(mat[:, 0:2])
```

## Filtrando com condições
A comparação gera Verdadeiro/Falso para cada elemento; colocada entre colchetes, mantém só os True:

💡  E = & · OU = | · NÃO = ~ · cada condição entre parênteses · é EXATAMENTE assim que o Pandas filtra linhas.

```python
precos = np.array([28.9, 8.5, 5.2, 18.9, 5.8])
estoque = np.array([40, 120, 80, 35, 60])

print(precos > 10)
print(precos[precos > 10])
print(estoque[estoque < 50])

barato_e_cheio = (precos < 10) & (estoque > 70)
print(barato_e_cheio)
print(precos[barato_e_cheio])
```

## Cuidado: fatia é uma "janela", não uma cópia
Uma fatia (slice) mostra os MESMOS dados do array original. Mudar a fatia muda o original!

💡  Quer mexer sem afetar o original? Use .copy(). Sem .copy(), a fatia é só uma "janela" para o mesmo dado.

```python
a = np.array([10, 20, 30, 40, 50])
fatia = a[1:4]
fatia[0] = 999
print(a)
print(fatia, end='\n\n')

b = np.array([10, 20, 30, 40, 50])
copia = b[1:4].copy()
copia[0] = 999
print(b)
print(copia)
```

## Contas elemento a elemento
As contas valem para TODOS os elementos, posição por posição. Nada de for:

💡  Os dois arrays precisam ter o mesmo tamanho (ou ser compatíveis, o broadcasting que vem logo adiante).

```python
a = np.array([2, 4, 6, 8])
b = np.array([1, 2, 3, 4])

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a ** 2)
print(a * 1.10)
```

## Um caso do Mercadinho: faturamento
Preço × quantidade, item a item, e depois um resumo:

💡  argmax() devolve a POSIÇÃO do maior valor (e argmin() a do menor). Ligue a posição ao produto: 0 = Arroz.

```python
precos = np.array([28.9, 8.5, 5.2, 18.9])
qtd = np.array([3, 10, 6, 2])

faturamento = np.round(precos * qtd, 2)
print(faturamento)
print('Total:', faturamento.sum())
print('Maior venda:', faturamento.max())
print('Posição da maior:', faturamento.argmax())
```

## Estatísticas rápidas
O array "se resume" em um número (ou em um número por linha/coluna):

💡  mean = média · median = mediana (valor do meio) · std = desvio padrão (o quanto os valores variam) · np.sqrt = raiz quadrada de cada um.

```python
notas = np.array([7.0, 8.5, 6.0, 9.5, 5.5])

print(notas.sum())
print(notas.mean())
print(np.median(notas))
print(notas.max(), notas.min())
print(notas.std().round(2))
print(notas.argmax(), notas.argmin())
print(np.sqrt(notas).round(2))
```

## Números aleatórios (random)
Gerar dados de teste e fazer simulações: um gerador com semente, para o resultado se repetir.

💡  seed fixa a "semente": com a mesma semente o resultado é sempre o mesmo · integers = inteiros · random = entre 0 e 1 · normal = distribuição normal · choice = sorteio.

```python
rng = np.random.default_rng(seed=42)

print(rng.integers(1, 7, size=10))
print(rng.random(3).round(3))
print(rng.normal(50, 10, size=5).round(1))

vendas_sim = rng.integers(80, 200, size=(3, 4))
print(vendas_sim)
print(rng.choice(['Ana', 'Marcos', 'Rafael'], size=2, replace=False))
```

## Excel × Pandas

## Dois personagens: Series e DataFrame
- **DataFrame** = a tabela inteira
- **Series** = uma coluna
- **index** = o rótulo (número) de cada linha

## Series: uma coluna com rótulos
Uma Series é UMA coluna: tem valores e tem rótulos (o índice). Como os preços de cada produto:

💡  Pense no índice como o nome de cada linha da planilha. Series aceita as mesmas contas do Numpy: precos * 1.10 reajusta todos.

```python
precos = pd.Series([28.9, 8.5, 5.2, 18.9],
                   index=['Arroz', 'Feijão', 'Macarrão', 'Café'])

print(precos, end='\n\n')
print(f"Preço do Feijão: R${precos['Feijão']}", end='\n\n')
print(f"Valores: {precos.values}", end='\n\n')
print(f"Índices: {precos.index.tolist()}", end='\n\n')
print(f"Média: {precos.mean()}, Máximo: {precos.max()}", end='\n\n')
print(f"Preços com aumento de 10%:\n{precos * 1.10}", end='\n\n')
```

## DataFrame: uma tabela completa
Um DataFrame é uma tabela com linhas e colunas. Cada coluna é uma Series e pode ter um tipo diferente:

💡  Chave do dicionário = nome da coluna · lista = valores · uma coluna nova nasce de uma conta com colunas existentes.

```python
mini = pd.DataFrame({
    'produto': ['Arroz', 'Feijão', 'Macarrão', 'Café'],
    'preco': [28.9, 8.5, 5.2, 18.9],
    'estoque': [40, 120, 80, 35]})
print(mini)

print(type(mini['preco']))

mini['valor_estoque'] = mini['preco'] * mini['estoque']
print(mini)
print(mini.shape)
```

## Index: o endereço de cada linha
Por padrão o índice é 0, 1, 2…; com set_index você troca pelo nome do produto e localiza linhas pelo nome:

💡  loc → pelo RÓTULO (nome) · iloc → pela POSIÇÃO (número) · reset_index() devolve o índice numérico.

```python
tabela = mini.set_index('produto')
print(tabela, end='\n\n')

print(tabela.loc['Café'], end='\n\n')
print(tabela.iloc[0], end='\n\n')
print(tabela.index.tolist(), end='\n\n')
print(tabela.reset_index().columns.tolist(), end='\n\n')
```

## Tipos de dados (dtype)
Cada coluna do DataFrame tem um tipo, e o tipo define o que dá para fazer com ela:

| dtype | O que guarda | Exemplo | Serve para |
|---|---|---|---|
| int64 | números inteiros | 40, 120 | contar, somar |
| float64 | números decimais | 28.9, 8.5 | preços, médias |
| object | texto | Arroz, Feijão | nomes, categorias |
| bool | verdadeiro / falso | True, False | filtros |
| datetime64 | datas e horas | 2026-08-01 | diferenças, meses |

⚠️  Tipo errado = cálculo errado. Um número guardado como texto (object) não soma. Confira com df.dtypes ou df.info().

## loc × iloc
- `loc` → pelo **nome** (rótulo)
- `iloc` → pela **posição** (número, começa em 0)

## loc: pelo nome

```python
p = produtos.set_index('produto')[['categoria', 'preco', 'estoque']]

print(p.loc['Café 500g'])           # a linha inteira
print(p.loc['Café 500g', 'preco'])  # uma célula
```

## iloc: pela posição

```python
print(p.iloc[3])        # 4ª linha (começa em 0!)
print(p.iloc[3, 1])     # linha 3, coluna 1

p.iloc[0:3]             # linhas 0, 1 e 2
```

## Filtrando: pergunta que vira Verdadeiro/Falso
Passo 1: `produtos['preco'] > 10` → True/False por linha.
Passo 2: `produtos[...]` mantém só as linhas True.

## Filtro simples

```python
produtos[produtos['preco'] > 10][['produto', 'categoria', 'preco']]
```

## Combinando condições
| Símbolo | Significa | Exemplo |
|---|---|---|
| `&` | E (as duas) | barato **E** mercearia |
| `\|` | OU (uma delas) | bebidas **OU** limpeza |
| `~` | NÃO | **não** é mercearia |

## Filtro com E (&)

```python
mercearia = produtos['categoria'] == 'Mercearia'
barato = produtos['preco'] < 10

produtos[mercearia & barato][['produto', 'preco']]
```

## Filtros com OU (|) e NÃO (~)

```python
bebida = produtos['categoria'] == 'Bebidas'
limpeza = produtos['categoria'] == 'Limpeza'

print(produtos[bebida | limpeza]['produto'].tolist())
print(produtos[~mercearia].shape[0], 'produtos não são de mercearia')
```

## Erros comuns nos filtros
1. Usar `and` no lugar de `&` → *ValueError: truth value of a Series is ambiguous*
2. Esquecer os parênteses: `df['a'] > 1 & df['b'] < 5`
3. Usar `=` no lugar de `==` → *SyntaxError*

## Atalhos de filtro: isin, between, contains

```python
print(produtos[produtos['categoria'].isin(['Bebidas', 'Limpeza'])]['produto'].tolist())
print(produtos[produtos['preco'].between(5, 10)]['produto'].tolist())
print(produtos[produtos['produto'].str.contains('kg')]['produto'].tolist())
```

## Groupby: dividir, calcular e juntar

2. **Calcular** (somar, contar, tirar média…)

## Ranking de categorias

```python
ranking = vendas.groupby('categoria')['valor_total'].sum()
ranking.sort_values(ascending=False).round(2)
```

## Contar e tirar média

```python
print(vendas.groupby('vendedor')['id_venda'].count())
print()
print(vendas.groupby('pagamento')['valor_total'].mean().round(2))
```

## agg: várias contas de uma vez

```python
resumo = vendas.groupby('categoria').agg(
    faturamento=('valor_total', 'sum'),
    ticket_medio=('valor_total', 'mean'),
    n_vendas=('id_venda', 'count')
).round(2)
resumo
```

# 🔗 Bloco 5 — Juntar tabelas

## Merge: o PROCV do Python
A tabela **vendas** só tem o `id_cliente`. Nome e cidade estão na tabela **clientes**. O `merge` cola as duas usando a **chave** em comum.

## merge na prática

```python
clientes = pd.read_csv('dados/clientes.csv')
completo = vendas.merge(clientes, on='id_cliente', how='left')
completo[['id_venda', 'id_cliente', 'nome', 'cidade', 'valor_total']].head(5)
```

## E o cliente que não está no cadastro?

```python
completo[completo['id_cliente'] == 99][['id_venda', 'id_cliente', 'nome', 'cidade', 'valor_total']].head(4)
```

## how=: quais linhas ficam?
- `inner` → só quem existe nas duas tabelas
- `left` → todas da esquerda ⭐
- `right` → todas da direita
- `outer` → tudo

## Contando as linhas de cada how

```python
for how in ['inner', 'left', 'right', 'outer']:
    n = vendas.merge(clientes, on='id_cliente', how=how).shape[0]
    print(how, '->', n, 'linhas')
```

## Respondendo ao dono: vendas por cidade

```python
completo['cidade'] = completo['cidade'].fillna('Sem cadastro')
completo.groupby('cidade')['valor_total'].sum().sort_values(ascending=False).round(2)
```
